```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from tabulate import tabulate

# Actual Dataset
data = {
    "Time (min)": [0, 3, 6, 9, 12],
    "Average Voltage (v)": [0.000, 18.240, 23.780, 25.207, 22.693],
    "Average Current (A)": [0.000, 0.483, 0.513, 0.560, 0.500],
    "Average Power (w)": [0.000, 8.782, 12.223, 14.116, 11.723],
    "Average T1 (°C)": [32.000, 105.667, 163.250, 169.333, 165.750],
    "Average T2 (°C)": [32.000, 43.000, 51.917, 58.833, 52.383],
    "Average ΔT (°C)": [0.000, 62.667, 111.333, 115.500, 113.367],
}

# DataFrame
df = pd.DataFrame(data)

# Print Table
print("---Maize Straw: Average Experimental Data Table ---")
print(tabulate(df, headers="keys", tablefmt="fancy_grid", showindex=False))

# Voltage, Current and Temperatures
time = df["Time (min)"]
voltage = df["Average Voltage (v)"]
current = df["Average Current (A)"]
t1 = df["Average T1 (°C)"]
t2 = df["Average T2 (°C)"]
delta_t = df["Average ΔT (°C)"]

fig, ax1 = plt.subplots(figsize=(9, 5))

# Y -axis (Voltage & Current )
(l1,) = ax1.plot(
    time,
    voltage,
    marker="s",
    linestyle="--",
    color="blue",
    label="Voltage (V)",
)
(l2,) = ax1.plot(
    time, current, marker="o", linestyle="-", color="red", label="Current (A)"
)

ax1.set_xlabel("Time (min)", fontweight="bold")
ax1.set_ylabel("Voltage (V) & Current (A)", color="black", fontweight="bold")
ax1.tick_params(axis="y", labelcolor="black")
ax1.grid(True, linestyle="--", alpha=0.5)

# Y-axis (Temperatures: T1, T2, ΔT )
ax2 = ax1.twinx()
(l3,) = ax2.plot(
    time,
    t1,
    marker="^",
    linestyle="-.",
    color="darkorange",
    label="T1 (°C)",
)
(l4,) = ax2.plot(
    time, t2, marker="x", linestyle=":", color="deepskyblue", label="T2 (°C)"
)
(l5,) = ax2.plot(
    time,
    delta_t,
    marker="d",
    linestyle="-.",
    color="green",
    label="ΔT (°C)",
)

ax2.set_ylabel("Temperatures & ΔT (°C)", color="green", fontweight="bold")
ax2.tick_params(axis="y", labelcolor="green")

# Legend
lines = [l1, l2, l3, l4, l5]
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc="upper left", fontsize=9)

plt.title(
    "Average: Voltage, Current & Temperatures Trends",
    fontsize=12,
    fontweight="bold",
)
plt.tight_layout()
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. Feature Selection & Target Definition
X = df[
    [
        "Time (min)",
        "Average Voltage (v)",
        "Average Current (A)",
        "Average T1 (°C)",
        "Average T2 (°C)",
        "Average ΔT (°C)",
    ]
]
y = df["Average Power (w)"]

# 2. Model Training (Using full dataset for small experimental trials to avoid nan)
lr_model = LinearRegression()
lr_model.fit(X, y)
y_pred_lr = lr_model.predict(X)

rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X, y)
y_pred_rf = rf_model.predict(X)


# 3. Evaluation Function
def evaluate_model(y_true, y_pred, model_name):
  mse = mean_squared_error(y_true, y_pred)
  rmse = np.sqrt(mse)
  mae = mean_absolute_error(y_true, y_pred)
  r2 = r2_score(y_true, y_pred)

  print(f"--- {model_name} Performance ---")
  print(f"RMSE : {rmse:.4f}")
  print(f"MAE  : {mae:.4f}")
  print(f"R²   : {r2:.4f}\n")


# Result Print
print("=== Task-2 Machine Learning Model Evaluation Results (Full Data Fit) ===")
evaluate_model(y, y_pred_lr, "Linear Regression")
evaluate_model(y, y_pred_rf, "Random Forest Regressor")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Dataset (From Task-1 / Task-2)
data = {
    "Time (min)": [0, 3, 6, 9, 12],
    "Average Voltage (v)": [0.000, 18.240, 23.780, 25.207, 22.693],
    "Average Current (A)": [0.000, 0.483, 0.513, 0.560, 0.500],
    "Average Power (w)": [0.000, 8.782, 12.223, 14.116, 11.723],
    "Average T1 (°C)": [32.000, 105.667, 163.250, 169.333, 165.750],
    "Average T2 (°C)": [32.000, 43.000, 51.917, 58.833, 52.383],
    "Average ΔT (°C)": [0.000, 62.667, 111.333, 115.500, 113.367],
}

df = pd.DataFrame(data)

# 2. Intelligent Monitoring / Anomaly Detection Algorithm
def detect_anomalies(df):
    anomalies = []
    for index, row in df.iterrows():
        # Criteria 1: High Temperature Threshold (T1 > 170 °C indicates overheating risk)
        # Criteria 2: Sudden Power Drop (Power < 5.0 W while ΔT is high > 100 °C indicates potential system fault)
        if row["Average T1 (°C)"] > 170.0:
            anomalies.append((row["Time (min)"], "Overheating Risk (T1 > 170°C)"))
        elif row["Average ΔT (°C)"] > 100.0 and row["Average Power (w)"] < 5.0:
            anomalies.append((row["Time (min)"], "Sudden Power Drop / Fault Detected"))

    return anomalies

detected_anomalies = detect_anomalies(df)
print("=== Intelligent Monitoring Alarm Log ===")
if detected_anomalies:
    for time_min, reason in detected_anomalies:
        print(f"[ALARM] Time: {time_min} min -> Anomaly: {reason}")
else:
    print("System Operating Normally. No Anomalies Detected.")

# 3. Visualization of Anomalies
plt.figure(figsize=(8, 4))
plt.plot(df["Time (min)"], df["Average Power (w)"], marker="o", color="tab:red", label="Power (W)")
plt.plot(df["Time (min)"], df["Average ΔT (°C)"], marker="d", linestyle="--", color="tab:green", label="ΔT (°C)")

# Highlight normal operation range and thresholds
plt.axhline(y=5.0, color="orange", linestyle=":", label="Min Safe Power Threshold (5W)")
plt.title("Task-4: Intelligent Monitoring & Anomaly Detection")
plt.xlabel("Time (min)")
plt.ylabel("Power (W) / ΔT (°C)")
plt.legend(loc="upper left")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
