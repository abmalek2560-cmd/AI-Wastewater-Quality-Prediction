```python

# DATA QUALITY AND MISSING VALUE AUDIT


import pandas as pd
import numpy as np

# Keep the original dataset unchanged
df_raw = df.copy()

print("=" * 60)
print("DATA QUALITY AUDIT")
print("=" * 60)

# 1. Dataset dimensions
print("\n1. DATASET DIMENSIONS")
print("Rows:", df_raw.shape[0])
print("Columns:", df_raw.shape[1])

# 2. Missing values
missing = pd.DataFrame({
    "Missing_Count": df_raw.isnull().sum(),
    "Missing_Percentage": (df_raw.isnull().mean() * 100).round(2)
})

missing = missing[missing["Missing_Count"] > 0]
missing = missing.sort_values("Missing_Percentage", ascending=False)

print("\n2. MISSING VALUES")
display(missing)

# 3. Zero values in numeric columns
numeric_df = df_raw.select_dtypes(include=np.number)

zero_summary = pd.DataFrame({
    "Zero_Count": (numeric_df == 0).sum(),
    "Zero_Percentage": ((numeric_df == 0).mean() * 100).round(2)
})

zero_summary = zero_summary.sort_values(
    "Zero_Percentage", ascending=False
)

print("\n3. ZERO VALUES IN NUMERIC COLUMNS")
display(zero_summary)

# 4. Negative values
negative_summary = (numeric_df < 0).sum()
negative_summary = negative_summary[negative_summary > 0]

print("\n4. NEGATIVE VALUES")
if negative_summary.empty:
    print("No negative values found in numeric columns.")
else:
    display(negative_summary.to_frame("Negative_Count"))

# 5. Infinite values
infinity_summary = pd.Series(
    np.isinf(numeric_df).sum(),
    name="Infinite_Count"
)

infinity_summary = infinity_summary[infinity_summary > 0]

print("\n5. INFINITE VALUES")
if infinity_summary.empty:
    print("No infinite values found.")
else:
    display(infinity_summary.to_frame())

# 6. Check potential prediction target: Effluent BOD
target = "방류_생화학적산소요구량(BOD)"

print("\n6. EFFLUENT BOD TARGET AUDIT")

if target in df_raw.columns:
    bod = pd.to_numeric(df_raw[target], errors="coerce")

    print("Total observations:", len(bod))
    print("Valid observations:", bod.notna().sum())
    print("Missing observations:", bod.isna().sum())
    print("Zero observations:", (bod == 0).sum())
    print("Negative observations:", (bod < 0).sum())

    print("\nEffluent BOD descriptive statistics:")
    display(bod.describe().to_frame("Effluent_BOD"))
else:
    print("Effluent BOD column not found.")

print("\nAUDIT COMPLETED.")
```

```python


#  EFFLUENT BOD VALIDATION


target = "방류_생화학적산소요구량(BOD)"
influent = "유입_생화학적산소요구량(BOD)"

# 1. Recheck missing values
print("=" * 60)
print("1. MISSING VALUE VERIFICATION")
print("=" * 60)

print("Missing in original df:", df[target].isna().sum())
print("Missing in df_raw:", df_raw[target].isna().sum())

bod_numeric = pd.to_numeric(df_raw[target], errors="coerce")

print("Missing after numeric conversion:", bod_numeric.isna().sum())
print("Blank strings:",
      df_raw[target].astype(str).str.strip().eq("").sum())

# 2. Inspect zero BOD observations
print("\n" + "=" * 60)
print("2. ZERO EFFLUENT BOD OBSERVATIONS")
print("=" * 60)

zero_rows = df_raw.loc[
    bod_numeric.eq(0),
    ["시도", "시설명", "시설용량", "유입하수량",
     influent, target]
].copy()

print("Number of zero observations:", len(zero_rows))
display(zero_rows)

# 3. Inspect unusually high BOD observations
print("\n" + "=" * 60)
print("3. HIGHEST EFFLUENT BOD OBSERVATIONS")
print("=" * 60)

high_rows = df_raw.loc[
    bod_numeric >= 20,
    ["시도", "시설명", "시설용량", "유입하수량",
     influent, target]
].copy()

high_rows = high_rows.sort_values(
    target, ascending=False
)

print("Observations with Effluent BOD >= 20:",
      len(high_rows))

display(high_rows.head(20))

# 4. Compare influent and effluent BOD
print("\n" + "=" * 60)
print("4. INFLUENT VS EFFLUENT BOD SUMMARY")
print("=" * 60)

display(
    df_raw[[influent, target]]
    .apply(pd.to_numeric, errors="coerce")
    .describe()
)

# 5. Check whether effluent BOD exceeds influent BOD
influent_numeric = pd.to_numeric(
    df_raw[influent], errors="coerce"
)

valid_pair = (
    influent_numeric.notna() &
    bod_numeric.notna()
)

print("\nValid influent-effluent pairs:", valid_pair.sum())
print(
    "Effluent BOD > Influent BOD:",
    (bod_numeric[valid_pair] >
     influent_numeric[valid_pair]).sum()
)

print("\nPHASE 4 VALIDATION COMPLETED.")
```

```python

# PHASE 5 — STEP 1
# Descriptive Statistics

columns = [
    "시설용량",
    "유입하수량",
    "유입_생화학적산소요구량(BOD)",
    "방류_생화학적산소요구량(BOD)"
]

eda_df = df_raw.copy()

display(
    eda_df[columns]
    .describe()
    .round(2)
)
```

```python

# PHASE 5 — STEP 2
# Calculate BOD Removal Efficiency

influent = "유입_생화학적산소요구량(BOD)"
effluent = "방류_생화학적산소요구량(BOD)"

valid = eda_df[influent] > 0

eda_df["BOD_Removal_Percentage"] = np.nan

eda_df.loc[valid, "BOD_Removal_Percentage"] = (
    (eda_df.loc[valid, influent] -
     eda_df.loc[valid, effluent])
    / eda_df.loc[valid, influent]
) * 100

print("Total observations:", len(eda_df))
print("Valid observations:", valid.sum())
print("Excluded zero-influent observations:", (~valid).sum())

display(
    eda_df["BOD_Removal_Percentage"]
    .describe()
    .round(2)
    .to_frame("BOD_Removal_Percentage")
)
```

```python

# PHASE 5 — STEP 5
# Treatment Method Analysis

method = "처리방법"
target = "방류_생화학적산소요구량(BOD)"

method_summary = (
    eda_df.groupby(method, dropna=False)
    .agg(
        Facility_Count=(target, "count"),
        Mean_Effluent_BOD=(target, "mean"),
        Median_Effluent_BOD=(target, "median"),
        Mean_BOD_Removal=(
            "BOD_Removal_Percentage", "mean"
        )
    )
    .sort_values(
        "Facility_Count",
        ascending=False
    )
)

display(method_summary.round(2))
```

```python
#  Energy Performance Analysis

self_suff = "에너지자립율"
self_prod = "에너지자체생산량"
total_energy = "연간총에너지사용량_계_(TOE)"
electricity = "연간총에너지사용량_총전력사용량(kWh)"
own_energy = "연간총에너지사용량_자체에너지사용량(TOE)"

energy_df = df_raw[
    [self_suff, self_prod, total_energy, electricity, own_energy]
].copy()

# Count facilities with positive values
print("ENERGY PERFORMANCE SUMMARY")
print("-" * 40)

for col in energy_df.columns:
    positive_count = (energy_df[col] > 0).sum()
    print(f"{col}: {positive_count} facilities with positive values")

# Statistics for facilities with positive energy self-sufficiency
positive_self = energy_df[energy_df[self_suff] > 0]

print("\nENERGY SELF-SUFFICIENCY (Positive Values Only)")
display(positive_self[self_suff].describe().round(2).to_frame())

# Check whether self-produced energy and own energy use are identical
same_values = (
    energy_df[self_prod].round(6) ==
    energy_df[own_energy].round(6)
)

print("\nSELF-PRODUCTION vs OWN ENERGY USE")
print("Identical rows:", same_values.sum())
print("Different rows:", (~same_values).sum())

# Energy consumption statistics
print("\nENERGY CONSUMPTION STATISTICS")
display(
    energy_df[[total_energy, electricity, self_prod]]
    .describe()
    .round(2)
)
```

```python
#  Validate Energy Self-Sufficiency

self_suff = "에너지자립율"
self_prod = "에너지자체생산량"
total_energy = "연간총에너지사용량_계_(TOE)"

validation_df = df_raw[
    [self_suff, self_prod, total_energy]
].copy()

# Calculate ratio only where total energy consumption is positive
valid = validation_df[total_energy] > 0

validation_df["Calculated_Self_Sufficiency"] = np.nan

validation_df.loc[valid, "Calculated_Self_Sufficiency"] = (
    validation_df.loc[valid, self_prod] /
    validation_df.loc[valid, total_energy]
) * 100

# Compare reported and calculated values
validation_df["Difference"] = (
    validation_df[self_suff] -
    validation_df["Calculated_Self_Sufficiency"]
)

print("ENERGY SELF-SUFFICIENCY VALIDATION")
print("----------------------------------")
print("Valid facilities:", valid.sum())

print("\nReported Self-Sufficiency:")
display(validation_df.loc[valid, self_suff].describe().round(2))

print("\nCalculated Self-Sufficiency:")
display(
    validation_df.loc[valid, "Calculated_Self_Sufficiency"]
    .describe().round(2)
)

print("\nDifference between reported and calculated values:")
display(validation_df.loc[valid, "Difference"].describe().round(2))

print(
    "\nFacilities with difference within 0.1 percentage point:",
    (validation_df.loc[valid, "Difference"].abs() <= 0.1).sum()
)
```

```python
#  Investigate Energy Data Mismatches

# Select facilities with differences greater than 0.1
mismatch_df = validation_df.loc[
    validation_df["Difference"].abs() > 0.1
].copy()

# Add original row index
mismatch_df["Original_Row_Index"] = mismatch_df.index

# Include facility identification columns if available
possible_id_columns = [
    "시도", "구군", "시설명", "사업장명",
    "소재지", "시설용량"
]

available_id_columns = [
    col for col in possible_id_columns
    if col in df_raw.columns
]

# Combine identification and energy columns
result = pd.concat(
    [
        df_raw.loc[mismatch_df.index, available_id_columns],
        mismatch_df[
            [
                "Original_Row_Index",
                self_suff,
                self_prod,
                total_energy,
                "Calculated_Self_Sufficiency",
                "Difference"
            ]
        ]
    ],
    axis=1
)

# Show largest discrepancies first
result["Absolute_Difference"] = result["Difference"].abs()

result = result.sort_values(
    "Absolute_Difference",
    ascending=False
)

print("Total mismatched facilities:", len(result))
print("\nTop 20 facilities with largest differences:")

display(result.head(20).round(2))
```

```python
#  Classify Energy Self-Sufficiency Mismatches

mismatch_df = validation_df.loc[
    validation_df["Difference"].abs() > 0.1
].copy()

# Classify mismatch magnitude
mismatch_df["Mismatch_Category"] = pd.cut(
    mismatch_df["Difference"].abs(),
    bins=[0.1, 1, 5, 10, float("inf")],
    labels=[
        "0.1–1 percentage point",
        "1–5 percentage points",
        "5–10 percentage points",
        "Above 10 percentage points"
    ],
    include_lowest=True
)

# Summarize mismatch categories
category_summary = (
    mismatch_df["Mismatch_Category"]
    .value_counts(sort=False)
    .rename_axis("Mismatch_Category")
    .reset_index(name="Facility_Count")
)

category_summary["Percentage"] = (
    category_summary["Facility_Count"] /
    len(mismatch_df) * 100
)

print("MISMATCH CATEGORY SUMMARY")
display(category_summary.round(2))

# Summary of mismatch direction
mismatch_df["Difference_Direction"] = np.where(
    mismatch_df["Difference"] > 0,
    "Reported value is higher",
    "Calculated value is higher"
)

print("\nMISMATCH DIRECTION")
display(
    mismatch_df["Difference_Direction"]
    .value_counts()
    .to_frame("Facility_Count")
)
```

```python
#  Energy Self-Sufficiency vs CO2 Emission

self_suff = "에너지자립율"
co2_col = "하수처리량당이산화탄소(CO2)배출량"

co2_energy_df = df_raw[
    [self_suff, co2_col]
].copy()

# Keep records with positive self-sufficiency
# and non-missing CO2 values
co2_energy_df = co2_energy_df[
    (co2_energy_df[self_suff] > 0) &
    (co2_energy_df[co2_col].notna())
]

print("Facilities analyzed:", len(co2_energy_df))

# Calculate correlations
from scipy.stats import pearsonr, spearmanr

pearson_r, pearson_p = pearsonr(
    co2_energy_df[self_suff],
    co2_energy_df[co2_col]
)

spearman_r, spearman_p = spearmanr(
    co2_energy_df[self_suff],
    co2_energy_df[co2_col]
)

print("\nCO2 AND ENERGY SELF-SUFFICIENCY CORRELATION")
print("-------------------------------------------")
print(f"Pearson r: {pearson_r:.3f}")
print(f"Pearson p-value: {pearson_p:.4f}")
print(f"Spearman correlation: {spearman_r:.3f}")
print(f"Spearman p-value: {spearman_p:.4f}")
```

```python
# TOC PERFORMANCE ANALYSIS

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

toc_in = "유입_총유기탄소(TOC)"
toc_out = "방류_총유기탄소(TOC)"

# Prepare data
toc_df = df_raw[[toc_in, toc_out]].copy()

toc_df[toc_in] = pd.to_numeric(toc_df[toc_in], errors="coerce")
toc_df[toc_out] = pd.to_numeric(toc_df[toc_out], errors="coerce")

toc_df = toc_df.dropna()

print("Total valid facilities:", len(toc_df))

# Descriptive statistics
print("\nInfluent TOC Statistics:")
print(toc_df[toc_in].describe())

print("\nEffluent TOC Statistics:")
print(toc_df[toc_out].describe())

# TOC removal percentage
valid = toc_df[toc_in] > 0

toc_df["TOC_Removal_Percentage"] = float("nan")

toc_df.loc[valid, "TOC_Removal_Percentage"] = (
    (toc_df.loc[valid, toc_in] - toc_df.loc[valid, toc_out])
    / toc_df.loc[valid, toc_in]
) * 100

print("\nValid facilities for TOC removal:", valid.sum())

print("\nTOC Removal Statistics:")
print(toc_df["TOC_Removal_Percentage"].describe())

# Compare influent and effluent TOC
plt.figure(figsize=(9, 5))

sns.histplot(
    toc_df[toc_in],
    color="blue",
    label="Influent TOC",
    kde=True,
    stat="density",
    alpha=0.35
)

sns.histplot(
    toc_df[toc_out],
    color="orange",
    label="Effluent TOC",
    kde=True,
    stat="density",
    alpha=0.35
)

plt.title("Influent vs Effluent TOC Distribution")
plt.xlabel("TOC")
plt.ylabel("Density")
plt.legend()
plt.tight_layout()
plt.show()

# TOC removal distribution
plt.figure(figsize=(9, 5))

sns.histplot(
    toc_df["TOC_Removal_Percentage"].dropna(),
    bins=40,
    kde=True
)

plt.title("TOC Removal Percentage Distribution")
plt.xlabel("TOC Removal (%)")
plt.ylabel("Number of Facilities")
plt.tight_layout()
plt.show()
```

```python
# TOC REMOVAL OUTLIER ANALYSIS

import matplotlib.pyplot as plt
import seaborn as sns

toc_removal = toc_df["TOC_Removal_Percentage"].dropna()

# IQR calculation
Q1 = toc_removal.quantile(0.25)
Q3 = toc_removal.quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = toc_removal[
    (toc_removal < lower_limit) |
    (toc_removal > upper_limit)
]

print("TOC Removal Outlier Analysis")
print("--------------------------------")
print("Valid facilities:", len(toc_removal))
print("Q1:", round(Q1, 3))
print("Q3:", round(Q3, 3))
print("IQR:", round(IQR, 3))
print("Lower limit:", round(lower_limit, 3))
print("Upper limit:", round(upper_limit, 3))
print("Potential outliers:", len(outliers))

print("\nOutlier values:")
print(outliers.sort_values())

# Boxplot
plt.figure(figsize=(9, 4))
sns.boxplot(x=toc_removal)
plt.title("TOC Removal Percentage - Outlier Detection")
plt.xlabel("TOC Removal (%)")
plt.tight_layout()
plt.show()
```

```python
#  BOD vs TOC COMPARATIVE ANALYSIS

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr, spearmanr

# Column names
bod_in = "유입_생화학적산소요구량(BOD)"
bod_out = "방류_생화학적산소요구량(BOD)"
toc_in = "유입_총유기탄소(TOC)"
toc_out = "방류_총유기탄소(TOC)"

# Prepare data
compare_df = df_raw[[bod_in, bod_out, toc_in, toc_out]].copy()

# Calculate removal percentages
compare_df["BOD_Removal"] = (
    (compare_df[bod_in] - compare_df[bod_out])
    / compare_df[bod_in].replace(0, float("nan"))
) * 100

compare_df["TOC_Removal"] = (
    (compare_df[toc_in] - compare_df[toc_out])
    / compare_df[toc_in].replace(0, float("nan"))
) * 100

# Keep facilities with both valid removal values
valid_df = compare_df.dropna(subset=["BOD_Removal", "TOC_Removal"])

print("Facilities with both valid measurements:", len(valid_df))

print("\nBOD Removal Statistics:")
print(valid_df["BOD_Removal"].describe())

print("\nTOC Removal Statistics:")
print(valid_df["TOC_Removal"].describe())

# Compare mean and median
print("\nComparative Summary:")
print(
    valid_df[["BOD_Removal", "TOC_Removal"]]
    .agg(["mean", "median", "std"])
)

# Correlation between BOD and TOC removal
pearson_r, pearson_p = pearsonr(
    valid_df["BOD_Removal"],
    valid_df["TOC_Removal"]
)

spearman_r, spearman_p = spearmanr(
    valid_df["BOD_Removal"],
    valid_df["TOC_Removal"]
)

print("\nCorrelation Results:")
print(f"Pearson r: {pearson_r:.4f}, p-value: {pearson_p:.6g}")
print(f"Spearman rho: {spearman_r:.4f}, p-value: {spearman_p:.6g}")

# Boxplot comparison
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=valid_df[["BOD_Removal", "TOC_Removal"]]
)

plt.title("BOD vs TOC Removal Performance")
plt.xlabel("Parameter")
plt.ylabel("Removal Percentage (%)")
plt.tight_layout()
plt.show()

# Scatter plot
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=valid_df,
    x="BOD_Removal",
    y="TOC_Removal",
    alpha=0.5
)

plt.title("BOD Removal vs TOC Removal")
plt.xlabel("BOD Removal (%)")
plt.ylabel("TOC Removal (%)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
